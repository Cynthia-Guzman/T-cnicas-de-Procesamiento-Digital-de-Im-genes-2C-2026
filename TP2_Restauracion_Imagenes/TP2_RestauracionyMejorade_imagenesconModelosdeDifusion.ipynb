### 2.3 Restauración de fotografías con Stable Diffusion x4

En esta sección se procesa la fotografía seleccionada mediante el modelo `stabilityai/stable-diffusion-x4-upscaler`, que permite ampliar cuatro veces su resolución.

Para controlar el consumo de memoria de la GPU, se limita el tamaño de entrada a 512 píxeles en su dimensión mayor. Además, se genera una ampliación bicúbica convencional para comparar ambos métodos.

Finalmente, se visualizan la fotografía original, la ampliación bicúbica y el resultado de Stable Diffusion x4, con el objetivo de observar las diferencias en nitidez, textura y reconstrucción de detalles.

````python

# ============================================================
# PASO 2.3 - RESTAURACIÓN DE FOTOGRAFÍAS CON STABLE DIFFUSION X4
# ============================================================

# Cargamos el modelo de superresolución y procesamos
# la fotografía seleccionada en la celda anterior.
#
# Se comparan tres imágenes:
# 1. Fotografía original
# 2. Ampliación convencional bicúbica x4
# 3. Restauración mediante Stable Diffusion x4

from PIL import Image
from diffusers import StableDiffusionUpscalePipeline
import matplotlib.pyplot as plt
import torch

# ------------------------------------------------------------
# 1. VERIFICACIÓN DE LA FOTOGRAFÍA Y LA GPU
# ------------------------------------------------------------

if "nueva_foto" not in globals() or nueva_foto is None:
    raise ValueError("Primero seleccioná una fotografía.")

if not torch.cuda.is_available():
    raise RuntimeError(
        "Se necesita una GPU CUDA para ejecutar Stable Diffusion x4."
    )

dispositivo = "cuda"
print("GPU:", torch.cuda.get_device_name(0))

# ------------------------------------------------------------
# 2. CARGA DEL MODELO STABLE DIFFUSION X4
# ------------------------------------------------------------

# Cargamos el modelo una sola vez por sesión.
# Utilizamos float16 para reducir el consumo de memoria.
# El modelo se ejecuta con descarga parcial a CPU
# para evitar superar la memoria de la Tesla T4.

if "pipe_ampliar" not in globals():

    print("Cargando Stable Diffusion x4...")

    pipe_ampliar = StableDiffusionUpscalePipeline.from_pretrained(
        "stabilityai/stable-diffusion-x4-upscaler",
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True
    )

    pipe_ampliar.enable_model_cpu_offload()

    print("Modelo cargado correctamente.")

else:
    print("Utilizando el modelo ya cargado.")

# ------------------------------------------------------------
# 3. PREPARACIÓN DE LA FOTOGRAFÍA
# ------------------------------------------------------------

# Limitamos la dimensión mayor a 512 píxeles
# para controlar el consumo de memoria de la GPU.

ancho, alto = nueva_foto.size
LIMITE = 512

if max(ancho, alto) <= LIMITE:
    entrada = nueva_foto.copy()
else:
    escala = LIMITE / max(ancho, alto)

    nuevo_tamano = (
        max(1, round(ancho * escala)),
        max(1, round(alto * escala))
    )

    entrada = nueva_foto.resize(
        nuevo_tamano,
        Image.Resampling.LANCZOS
    )

# Ajustamos las dimensiones a múltiplos de cuatro.

ancho_ajustado = max(4, round(entrada.width / 4) * 4)
alto_ajustado = max(4, round(entrada.height / 4) * 4)

if entrada.size != (ancho_ajustado, alto_ajustado):
    entrada = entrada.resize(
        (ancho_ajustado, alto_ajustado),
        Image.Resampling.LANCZOS
    )

# ------------------------------------------------------------
# 4. AMPLIACIÓN CONVENCIONAL BICÚBICA
# ------------------------------------------------------------

# Generamos una ampliación x4 sin inteligencia artificial
# para utilizarla como referencia de comparación.

bicubica_6 = entrada.resize(
    (entrada.width * 4, entrada.height * 4),
    Image.Resampling.BICUBIC
)

# ------------------------------------------------------------
# 5. RESTAURACIÓN MEDIANTE STABLE DIFFUSION X4
# ------------------------------------------------------------

# El modelo amplía la fotografía cuatro veces.
# El prompt busca preservar la anatomía y los rasgos
# originales del perro, aunque no garantiza fidelidad.

print("Procesando fotografía con Stable Diffusion x4...")

with torch.inference_mode():

    resultado_6 = pipe_ampliar(
        prompt=(
            "High quality restoration of the input photograph, "
            "preserve the original face shape, muzzle proportions, "
            "anatomy and identity, natural photographic details"
        ),
        image=entrada,
        num_inference_steps=20,
        noise_level=5,
        guidance_scale=2.0
    )

difusion_6 = resultado_6.images[0].convert("RGB")

# ------------------------------------------------------------
# 6. PREPARACIÓN DE LA COMPARACIÓN
# ------------------------------------------------------------

# Ajustamos la fotografía original al tamaño de salida
# para visualizar las tres imágenes con iguales dimensiones.

original_6 = nueva_foto.resize(
    difusion_6.size,
    Image.Resampling.LANCZOS
)

print("\nFotografía original:", nueva_foto.size)
print("Entrada al modelo:", entrada.size)
print("Salida de Stable Diffusion:", difusion_6.size)

# ------------------------------------------------------------
# 7. VISUALIZACIÓN DE RESULTADOS
# ------------------------------------------------------------

fig, ejes = plt.subplots(1, 3, figsize=(16, 6))

versiones = [
    ("Fotografía original", original_6),
    ("Ampliación bicúbica x4", bicubica_6),
    ("Stable Diffusion x4", difusion_6)
]

for eje, (titulo, imagen) in zip(ejes, versiones):
    eje.imshow(imagen)
    eje.set_title(titulo)
    eje.axis("off")

plt.tight_layout()
plt.show()
````

### 2.4 Reducción de ruido de la imagen restaurada

Después de ampliar la fotografía mediante Stable Diffusion x4, se aplica una reducción de ruido utilizando el algoritmo *Non-Local Means* de OpenCV (`fastNlMeansDenoisingColored`).

Se utiliza una intensidad baja (`h=3`) para suavizar pequeñas irregularidades sin perder demasiados detalles de la imagen, especialmente en el pelaje y los rasgos del perro.

Finalmente, se comparan la fotografía original, la restauración mediante Stable Diffusion x4 y la versión con reducción de ruido para observar las diferencias visuales.

````python

# ============================================================
# PASO 2.4 - REDUCCIÓN DE RUIDO DE LA IMAGEN RESTAURADA
# ============================================================

# Aplicamos una técnica de reducción de ruido a la imagen
# generada por Stable Diffusion x4.
#
# Utilizamos Non-Local Means de OpenCV con intensidad h=3
# para suavizar pequeñas irregularidades de color y textura.
#
# Comparamos la fotografía original, la restauración con IA
# y el resultado posterior a la reducción de ruido.

import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ------------------------------------------------------------
# 1. PREPARACIÓN DE LA IMAGEN
# ------------------------------------------------------------

# Convertimos la imagen restaurada a un arreglo NumPy.
imagen = np.array(difusion_6.convert("RGB"))

# OpenCV utiliza el orden de canales BGR.
imagen_bgr = cv2.cvtColor(imagen, cv2.COLOR_RGB2BGR)

# ------------------------------------------------------------
# 2. APLICACIÓN DE REDUCCIÓN DE RUIDO
# ------------------------------------------------------------

# Aplicamos Non-Local Means con una intensidad suave.
# h y hColor controlan la reducción de ruido.
# Los otros parámetros determinan el tamaño de las
# ventanas utilizadas para analizar la imagen.

resultado_bgr = cv2.fastNlMeansDenoisingColored(
    imagen_bgr,
    None,
    h=3,
    hColor=3,
    templateWindowSize=7,
    searchWindowSize=21
)

# ------------------------------------------------------------
# 3. CONVERSIÓN DEL RESULTADO
# ------------------------------------------------------------

# Convertimos nuevamente a RGB para visualizar la imagen
# y utilizarla en las siguientes etapas del proyecto.

sin_ruido = Image.fromarray(
    cv2.cvtColor(resultado_bgr, cv2.COLOR_BGR2RGB)
)

# ------------------------------------------------------------
# 4. COMPARACIÓN VISUAL
# ------------------------------------------------------------

fig, ejes = plt.subplots(1, 3, figsize=(16, 6))

versiones = [
    ("Fotografía original", original_6),
    ("Stable Diffusion x4", difusion_6),
    ("Difusión + reducción de ruido (h=3)", sin_ruido)
]

for eje, (titulo, foto) in zip(ejes, versiones):
    eje.imshow(foto)
    eje.set_title(titulo)
    eje.axis("off")

plt.tight_layout()
plt.show()
````

````python

# ============================================================
# PASO 2.5.2 - ANÁLISIS COMPARATIVO CON SIGLIP
# ============================================================

import torch
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. DESCRIPCIONES PARA EL ANÁLISIS
# ------------------------------------------------------------

# SigLIP compara cada fotografía con cuatro descripciones
# relacionadas con la apariencia y la calidad visual.

descripciones = [
    "A clear, detailed photograph of a dog with natural fur texture.",
    "A blurry photograph of a dog with little visible detail.",
    "A noisy, grainy photograph of a dog.",
    "An artificial-looking photograph of a dog with distorted details."
]

etiquetas = [
    "Detalles naturales",
    "Desenfoque",
    "Ruido",
    "Apariencia artificial"
]

# ------------------------------------------------------------
# 2. FOTOGRAFÍAS A COMPARAR
# ------------------------------------------------------------

imagenes = [
    original_6.convert("RGB"),
    difusion_6.convert("RGB")
]

nombres = [
    "Fotografía original",
    "Stable Diffusion x4"
]

resultados_siglip = {}

# ------------------------------------------------------------
# 3. ANÁLISIS CON SIGLIP
# ------------------------------------------------------------

for nombre, imagen in zip(nombres, imagenes):

    # Preparar imagen y descripciones para el modelo.
    entradas = procesador_visual(
        text=descripciones,
        images=imagen,
        padding="max_length",
        return_tensors="pt"
    )

    # Calcular compatibilidad entre imagen y texto.
    with torch.inference_mode():
        salida = siglip(**entradas)

        puntuaciones = torch.sigmoid(
            salida.logits_per_image
        )[0]

    # Guardar los resultados para compararlos.
    resultados_siglip[nombre] = (
        puntuaciones.cpu().numpy()
    )

    # Mostrar las puntuaciones numéricas.
    print(f"\n{nombre}")

    for etiqueta, puntuacion in zip(
        etiquetas, puntuaciones
    ):
        print(f"{etiqueta}: {puntuacion.item():.3f}")


# ------------------------------------------------------------
# 4. GRÁFICO COMPARATIVO DE SIGLIP
# ------------------------------------------------------------

# Utilizamos las puntuaciones calculadas anteriormente.
# Ajustamos automáticamente la escala vertical para que
# las diferencias entre ambas fotografías sean visibles.

posiciones = np.arange(len(etiquetas))
ancho_barra = 0.35

original = resultados_siglip["Fotografía original"]
restaurada = resultados_siglip["Stable Diffusion x4"]

fig, eje = plt.subplots(figsize=(12, 6))

barras_original = eje.bar(
    posiciones - ancho_barra / 2,
    original,
    ancho_barra,
    label="Fotografía original"
)

barras_restaurada = eje.bar(
    posiciones + ancho_barra / 2,
    restaurada,
    ancho_barra,
    label="Stable Diffusion x4"
)

# Mostrar el valor encima de cada barra.
eje.bar_label(barras_original, fmt="%.4f", padding=3)
eje.bar_label(barras_restaurada, fmt="%.4f", padding=3)

# Ajustar la escala al valor máximo observado.
maximo = max(np.max(original), np.max(restaurada))
eje.set_ylim(0, maximo * 1.25 if maximo > 0 else 1)

eje.set_xticks(posiciones)
eje.set_xticklabels(etiquetas)
eje.set_ylabel("Puntuación de compatibilidad")
eje.set_title("Comparación visual mediante SigLIP")
eje.legend()

plt.tight_layout()
plt.show()

````

### 2.6 Mapa de diferencias: identificación de detalles reconstruidos

Los modelos de difusión pueden generar detalles que no estaban presentes en la fotografía original. Por este motivo, es importante analizar las modificaciones introducidas durante la restauración.

Se utiliza OpenCV y el operador Laplaciano para comparar los detalles finos de la ampliación bicúbica con los de la imagen restaurada mediante Stable Diffusion x4.

Las regiones donde se detecta un incremento significativo de detalle se resaltan en rojo sobre la fotografía restaurada.

**Interpretación:** las zonas rojas indican un aumento de detalle fino respecto de la ampliación convencional. No demuestran necesariamente que el modelo haya inventado información, pero permiten identificar regiones que requieren una revisión visual, especialmente en los rasgos del perro.

````python

# ============================================================
# PASO 2.6 - MAPA DE DIFERENCIAS
# ============================================================

# Identificamos las regiones donde Stable Diffusion x4
# presenta un incremento de detalle fino respecto
# de una ampliación bicúbica convencional.
#
# Las zonas detectadas se resaltan en rojo.
# Este mapa no demuestra que existan alucinaciones:
# señala regiones que requieren inspección visual.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. PREPARACIÓN DE LAS IMÁGENES
# ------------------------------------------------------------

# Utilizamos la ampliación bicúbica ya calculada.
original = np.array(bicubica_6.convert("RGB"))

# Imagen restaurada mediante Stable Diffusion x4.
restaurada = np.array(difusion_6.convert("RGB"))

# ------------------------------------------------------------
# 2. CONVERSIÓN A ESCALA DE GRISES
# ------------------------------------------------------------

gris_original = cv2.cvtColor(
    original,
    cv2.COLOR_RGB2GRAY
)

gris_ia = cv2.cvtColor(
    restaurada,
    cv2.COLOR_RGB2GRAY
)

# ------------------------------------------------------------
# 3. DETECCIÓN DE DETALLES FINOS
# ------------------------------------------------------------

# El operador Laplaciano permite identificar cambios
# locales de intensidad asociados con bordes y texturas.

detalle_original = np.abs(
    cv2.Laplacian(gris_original, cv2.CV_32F)
)

detalle_ia = np.abs(
    cv2.Laplacian(gris_ia, cv2.CV_32F)
)

# ------------------------------------------------------------
# 4. CÁLCULO DEL INCREMENTO DE DETALLE
# ------------------------------------------------------------

# Conservamos únicamente los valores positivos:
# regiones donde la imagen restaurada presenta
# mayor respuesta del Laplaciano.

incremento = np.maximum(
    detalle_ia - detalle_original,
    0
)

# Seleccionamos las zonas con incrementos más altos.
umbral = max(5, np.percentile(incremento, 95))

mascara = incremento > umbral

# Ampliamos ligeramente las regiones detectadas
# para facilitar su visualización.

mascara = mascara.astype(np.uint8) * 255

mascara = cv2.dilate(
    mascara,
    np.ones((3, 3), np.uint8)
)

# ------------------------------------------------------------
# 5. SUPERPOSICIÓN DEL MAPA EN COLOR ROJO
# ------------------------------------------------------------

resultado = restaurada.copy()

rojo = np.zeros_like(resultado)
rojo[:, :] = [255, 0, 0]

# Mezclamos el rojo con la fotografía restaurada.
mezcla = cv2.addWeighted(
    resultado,
    0.55,
    rojo,
    0.45,
    0
)

# Aplicamos el color solamente a las regiones detectadas.
resultado[mascara > 0] = mezcla[mascara > 0]

# ------------------------------------------------------------
# 6. VISUALIZACIÓN COMPARATIVA
# ------------------------------------------------------------

fig, ejes = plt.subplots(1, 3, figsize=(17, 6))

versiones = [
    ("Fotografía original", nueva_foto),
    ("Restauración con Stable Diffusion x4", difusion_6),
    ("Mapa de incremento de detalle fino", resultado)
]

for eje, (titulo, imagen) in zip(ejes, versiones):
    eje.imshow(imagen)
    eje.set_title(titulo)
    eje.axis("off")

plt.tight_layout()
plt.show()
````

### Interpretación del mapa de incremento de detalle fino

En el mapa podemos observar que las zonas marcadas en rojo se concentran principalmente en el fondo de la fotografía, especialmente en el pasto, y también en algunas partes del perro, como los ojos, el hocico, las orejas y el collar.

Esto indica que Stable Diffusion x4 generó más detalles finos en esas zonas en comparación con la ampliación bicúbica.

Sin embargo, que aparezcan más detalles no significa que sean reales. Algunos pueden haber sido reconstruidos o inventados por la inteligencia artificial.

Por eso, este mapa nos ayuda a identificar dónde hubo un aumento de detalle, pero no permite comprobar si esos detalles estaban presentes en la fotografía original.

En la comparación visual se observa que la interpolación bicúbica produce una imagen más borrosa, mientras que Stable Diffusion genera mayor nitidez y detalles aparentes en el rostro y el pelaje del perro.

````python

# ============================================================
# PASO 2.7.4 - EVALUACIÓN CUANTITATIVA CON PSNR Y SSIM
# ============================================================

# Comparamos la ampliación bicúbica y Stable Diffusion x4
# con la fotografía original de alta resolución.
#
# PSNR: cuanto mayor, menor error de píxeles.
# SSIM: cuanto más cercano a 1, mayor similitud estructural.

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from skimage.metrics import (
    peak_signal_noise_ratio,
    structural_similarity
)

# ------------------------------------------------------------
# 1. PREPARAR IMÁGENES CON LA MISMA RESOLUCIÓN
# ------------------------------------------------------------

# La referencia es la fotografía original.
referencia = referencia_hr.convert("RGB")

# La ampliación bicúbica ya tiene el tamaño correcto.
bicubica_eval = bicubica_prueba.convert("RGB")

# Ajustamos la salida de Stable Diffusion al tamaño
# de referencia para permitir la comparación.
difusion_eval = difusion_prueba.resize(
    referencia.size,
    Image.Resampling.LANCZOS
)

# ------------------------------------------------------------
# 2. CONVERTIR LAS IMÁGENES A NUMPY
# ------------------------------------------------------------

ref = np.array(referencia)
bic = np.array(bicubica_eval)
dif = np.array(difusion_eval)

# ------------------------------------------------------------
# 3. CALCULAR PSNR Y SSIM
# ------------------------------------------------------------

psnr_bicubica = peak_signal_noise_ratio(
    ref, bic, data_range=255
)

psnr_difusion = peak_signal_noise_ratio(
    ref, dif, data_range=255
)

ssim_bicubica = structural_similarity(
    ref, bic, channel_axis=2, data_range=255
)

ssim_difusion = structural_similarity(
    ref, dif, channel_axis=2, data_range=255
)

# ------------------------------------------------------------
# 4. MOSTRAR LOS RESULTADOS
# ------------------------------------------------------------

print("EVALUACIÓN DE FIDELIDAD")
print("-" * 45)

print(
    f"Ampliación bicúbica: "
    f"PSNR = {psnr_bicubica:.2f} dB | "
    f"SSIM = {ssim_bicubica:.4f}"
)

print(
    f"Stable Diffusion x4: "
    f"PSNR = {psnr_difusion:.2f} dB | "
    f"SSIM = {ssim_difusion:.4f}"
)

# ------------------------------------------------------------
# 5. GRÁFICOS COMPARATIVOS
# ------------------------------------------------------------

fig, ejes = plt.subplots(1, 2, figsize=(12, 5))

metodos = ["Bicúbica", "Stable Diffusion x4"]

ejes[0].bar(
    metodos,
    [psnr_bicubica, psnr_difusion]
)
ejes[0].set_title("PSNR")
ejes[0].set_ylabel("dB")

ejes[1].bar(
    metodos,
    [ssim_bicubica, ssim_difusion]
)
ejes[1].set_title("SSIM")
ejes[1].set_ylabel("Similitud estructural")
ejes[1].set_ylim(0, 1)

plt.tight_layout()
plt.show()
````

## 2.9 Determinación del umbral mínimo de resolución para una restauración confiable

En las pruebas anteriores observamos que Stable Diffusion x4 puede mejorar la nitidez de una fotografía, pero también generar detalles que no coinciden con la imagen original.

En particular, al reducir la fotografía del pug a 64 × 64 píxeles, el modelo produjo deformaciones importantes en los ojos, el hocico y las texturas del pelaje. En cambio, con una resolución de 128 × 128 píxeles, los rasgos del animal se conservaron mejor.

A partir de estos resultados, nos planteamos una nueva pregunta: ¿cuál es la resolución mínima que debería tener una fotografía para reducir el riesgo de que la inteligencia artificial invente o modifique detalles del perro?

Para investigarlo, realizaremos nuevas pruebas con resoluciones intermedias de 80 × 80, 96 × 96 y 112 × 112 píxeles, utilizando la misma fotografía y manteniendo los parámetros de Stable Diffusion x4.

Evaluaremos los resultados mediante las métricas PSNR y SSIM, junto con una comparación visual de los rasgos del animal.

El objetivo es identificar un umbral mínimo de resolución recomendable que podamos incorporar a nuestra aplicación como una medida de prevención.

De esta manera, cuando un usuario cargue una fotografía con una resolución inferior al umbral establecido, la aplicación podrá mostrar una advertencia indicando que la IA podría generar detalles artificiales o modificar características importantes para identificar al perro.

Este umbral será una recomendación experimental basada en las pruebas realizadas, no una garantía absoluta de fidelidad. Además, deberá validarse con otras fotografías, ya que la calidad de una imagen también depende de factores como el desenfoque, el ruido y la compresión.

Con esta evaluación buscamos que nuestra aplicación no solo mejore la calidad visual de las fotografías de perros de refugio, sino que también informe sobre las limitaciones de la restauración y promueva un uso responsable de la inteligencia artificial.

````python

# ============================================================
# PASO 2.9.1 - EVALUACIÓN DEL UMBRAL DE RESOLUCIÓN
# ============================================================

# Seleccionamos una fotografía desde la subcarpeta
# Prueba_Umbral de la carpeta Datos.
# Se prueban 80 x 80, 96 x 96 y 112 x 112 píxeles.
# Los resultados se conservan por fotografía.

import os
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from PIL import Image
from IPython.display import display, clear_output
from skimage.metrics import (
    peak_signal_noise_ratio,
    structural_similarity
)

# ------------------------------------------------------------
# 1. BUSCAR FOTOGRAFÍAS EN PRUEBA_UMBRAL
# ------------------------------------------------------------

# La carpeta DATOS original no se modifica.

DATOS_UMBRAL = os.path.join(DATOS, "Prueba_Umbral")

if not os.path.isdir(DATOS_UMBRAL):
    raise FileNotFoundError(
        f"No se encontró la carpeta: {DATOS_UMBRAL}"
    )

archivos_umbral = sorted([
    archivo
    for archivo in os.listdir(DATOS_UMBRAL)
    if archivo.lower().endswith((".jpg", ".jpeg", ".png"))
])

# Conservamos los resultados durante la sesión.
if "historial_umbral" not in globals():
    historial_umbral = {}

selector_umbral = widgets.Dropdown(
    options=[("Seleccioná una fotografía", None)] +
            [(nombre, nombre) for nombre in archivos_umbral],
    value=None,
    description="Fotografía:",
    style={"description_width": "initial"}
)

boton_umbral = widgets.Button(
    description="Evaluar fotografía",
    button_style="primary",
    icon="play"
)

salida_umbral = widgets.Output()

# ------------------------------------------------------------
# 2. FUNCIÓN DE EVALUACIÓN
# ------------------------------------------------------------

def ejecutar_prueba_umbral(boton):

    with salida_umbral:

        clear_output(wait=True)

        nombre = selector_umbral.value

        if nombre is None:
            print("Primero seleccioná una fotografía.")
            return

        # Cargar fotografía desde Prueba_Umbral.
        ruta_fotografia = os.path.join(
            DATOS_UMBRAL, nombre
        )

        fotografia = Image.open(
            ruta_fotografia
        ).convert("RGB")

        ancho, alto = fotografia.size

        print(f"Fotografía seleccionada: {nombre}")
        print(f"Resolución original: {ancho} x {alto} píxeles")

        # No se bloquea la evaluación por resolución.
        # Se informa la limitación cuando corresponde.

        if min(ancho, alto) < 512:

            print(
                "\nADVERTENCIA: la fotografía original "
                "tiene menos de 512 píxeles en uno de sus lados."
            )

            print(
                "Se ampliará para construir la referencia "
                "de 512 x 512 píxeles."
            )

            print(
                "Las métricas PSNR y SSIM deberán interpretarse "
                "con precaución porque la referencia ampliada "
                "no contiene detalles originales adicionales."
            )

        # ----------------------------------------------------
        # PREPARAR REFERENCIA
        # ----------------------------------------------------

        # Recortar un cuadrado centrado.
        lado = min(ancho, alto)

        izquierda = (ancho - lado) // 2
        arriba = (alto - lado) // 2

        recorte = fotografia.crop((
            izquierda,
            arriba,
            izquierda + lado,
            arriba + lado
        ))

        referencia = recorte.resize(
            (512, 512),
            Image.Resampling.LANCZOS
        )

        # ----------------------------------------------------
        # 3. EVALUAR RESOLUCIONES
        # ----------------------------------------------------

        resoluciones = [80, 96, 112]

        resultados = []
        imagenes = {}

        for resolucion in resoluciones:

            print(
                f"\nEvaluando {nombre}: "
                f"{resolucion} x {resolucion}..."
            )

            # Reducir resolución.
            imagen_lr = referencia.resize(
                (resolucion, resolucion),
                Image.Resampling.LANCZOS
            )

            # Interpolación bicúbica x4.
            bicubica = imagen_lr.resize(
                (resolucion * 4, resolucion * 4),
                Image.Resampling.BICUBIC
            )

            # Restauración mediante Stable Diffusion x4.
            with torch.inference_mode():

                resultado = pipe_ampliar(
                    prompt=(
                        "High quality photograph of a dog, "
                        "preserve the original face shape, "
                        "eyes, muzzle, fur colors, anatomy "
                        "and identity, natural photographic details"
                    ),
                    image=imagen_lr,
                    num_inference_steps=20,
                    noise_level=5,
                    guidance_scale=2.0
                )

            difusion = resultado.images[0].convert("RGB")

            # ------------------------------------------------
            # AJUSTAR IMÁGENES PARA COMPARAR MÉTRICAS
            # ------------------------------------------------

            bic_eval = bicubica.resize(
                referencia.size,
                Image.Resampling.LANCZOS
            )

            dif_eval = difusion.resize(
                referencia.size,
                Image.Resampling.LANCZOS
            )

            ref = np.array(referencia)
            bic = np.array(bic_eval)
            dif = np.array(dif_eval)

            # ------------------------------------------------
            # CALCULAR PSNR Y SSIM
            # ------------------------------------------------

            resultados.append({

                "Resolución": resolucion,

                "PSNR bicúbica": peak_signal_noise_ratio(
                    ref,
                    bic,
                    data_range=255
                ),

                "PSNR IA": peak_signal_noise_ratio(
                    ref,
                    dif,
                    data_range=255
                ),

                "SSIM bicúbica": structural_similarity(
                    ref,
                    bic,
                    channel_axis=2,
                    data_range=255
                ),

                "SSIM IA": structural_similarity(
                    ref,
                    dif,
                    channel_axis=2,
                    data_range=255
                )

            })

            imagenes[resolucion] = {
                "bicubica": bicubica.copy(),
                "difusion": difusion.copy()
            }

        # ----------------------------------------------------
        # 4. GUARDAR RESULTADOS POR FOTOGRAFÍA
        # ----------------------------------------------------

        tabla = pd.DataFrame(resultados)

        historial_umbral[nombre] = {
            "referencia": referencia.copy(),
            "resolucion_original": (ancho, alto),
            "resultados": tabla.copy(),
            "imagenes": imagenes
        }

        # ----------------------------------------------------
        # 5. MOSTRAR TABLA DE RESULTADOS
        # ----------------------------------------------------

        print("\nRESULTADOS:", nombre)

        display(tabla.round(4))

        # ----------------------------------------------------
        # 6. MOSTRAR COMPARACIONES VISUALES
        # ----------------------------------------------------

        fig, ejes = plt.subplots(
            len(resoluciones),
            3,
            figsize=(13, 12)
        )

        for i, resolucion in enumerate(resoluciones):

            versiones = [
                referencia,
                imagenes[resolucion]["bicubica"],
                imagenes[resolucion]["difusion"]
            ]

            titulos = [
                f"Referencia - {resolucion} px",
                "Bicúbica x4",
                "Stable Diffusion x4"
            ]

            for j, (imagen, titulo) in enumerate(
                zip(versiones, titulos)
            ):

                ejes[i, j].imshow(imagen)
                ejes[i, j].set_title(titulo)
                ejes[i, j].axis("off")

        plt.tight_layout()
        plt.show()

        print(
            "\nEvaluación finalizada. "
            "Resultados conservados para:", nombre
        )

# ------------------------------------------------------------
# 7. MOSTRAR SELECTOR Y BOTÓN
# ------------------------------------------------------------

boton_umbral.on_click(ejecutar_prueba_umbral)

display(
    widgets.VBox([
        selector_umbral,
        boton_umbral,
        salida_umbral
    ])
)


````

### PASO 2.9.2 - CONCLUSIÓN Y DEFINICIÓN DEL UMBRAL MÍNIMO DE RESOLUCIÓN

## 3.3 Integración del procesamiento de imágenes

Se integra la interfaz de Gradio con los métodos de restauración desarrollados anteriormente.

Al cargar una fotografía y presionar el botón de restauración, la aplicación realizará una ampliación bicúbica, una restauración mediante Stable Diffusion x4 y una reducción de ruido utilizando OpenCV.

Los resultados se mostrarán junto con la fotografía original para facilitar su comparación.

````python

# ============================================================
# PASO 3.3 - FUNCIÓN DE RESTAURACIÓN PARA GRADIO
# ============================================================

import numpy as np
import cv2
from PIL import Image

def restaurar_fotografia(foto):

    # Comprobar que se haya cargado una fotografía.
    if foto is None:
        raise gr.Error("Primero cargá una fotografía.")

    foto = foto.convert("RGB")

    # --------------------------------------------------------
    # 1. PREPARAR LA IMAGEN PARA STABLE DIFFUSION
    # --------------------------------------------------------

    # Limitar el tamaño de entrada para trabajar con la GPU.
    escala = min(1, 512 / max(foto.size))

    ancho = max(4, int(foto.width * escala) // 4 * 4)
    alto = max(4, int(foto.height * escala) // 4 * 4)

    entrada = foto.resize(
        (ancho, alto),
        Image.Resampling.LANCZOS
    )

    # --------------------------------------------------------
    # 2. AMPLIACIÓN BICÚBICA
    # --------------------------------------------------------

    bicubica = entrada.resize(
        (ancho * 4, alto * 4),
        Image.Resampling.BICUBIC
    )

    # --------------------------------------------------------
    # 3. RESTAURACIÓN CON STABLE DIFFUSION X4
    # --------------------------------------------------------

    prompt = (
        "High quality restoration of the input photograph, "
        "preserve the original face shape, muzzle proportions, "
        "anatomy and identity, natural photographic details"
    )

    with torch.inference_mode():
        restaurada = pipe_ampliar(
            prompt=prompt,
            image=entrada,
            num_inference_steps=20,
            noise_level=5,
            guidance_scale=2.0
        ).images[0]

    # --------------------------------------------------------
    # 4. REDUCCIÓN DE RUIDO CON OPENCV
    # --------------------------------------------------------

    imagen_bgr = cv2.cvtColor(
        np.array(restaurada),
        cv2.COLOR_RGB2BGR
    )

    filtrada_bgr = cv2.fastNlMeansDenoisingColored(
        imagen_bgr,
        None,
        h=3,
        hColor=3,
        templateWindowSize=7,
        searchWindowSize=21
    )

    sin_ruido = Image.fromarray(
        cv2.cvtColor(filtrada_bgr, cv2.COLOR_BGR2RGB)
    )

    # Devolver las tres imágenes a Gradio.
    return foto, bicubica, sin_ruido
````

````python


# ============================================================
# PASO 3.4 - FUNCIÓN DE DESCARGA
# ============================================================

import os
import tempfile
from PIL import Image

def preparar_descarga(imagen_sin_ruido):

    # Comprobar que exista una fotografía restaurada.
    if imagen_sin_ruido is None:
        raise gr.Error("Primero restaurá una fotografía.")

    # Guardar únicamente el resultado final:
    # Stable Diffusion x4 + OpenCV.
    nombre = "fotografia_restaurada.png"

    ruta = os.path.join(
        tempfile.mkdtemp(),
        nombre
    )

    imagen_sin_ruido.save(ruta)

    return ruta

````

## 3.5 Integración del mapa de diferencias y evaluación con SigLIP

Se incorporan dos herramientas de análisis para complementar la restauración de fotografías de animales y evaluar los cambios producidos durante la ampliación.

El **mapa de diferencias** permite identificar las zonas donde Stable Diffusion x4 genera mayores aumentos de detalle respecto de la ampliación bicúbica. Estas zonas se resaltan en rojo para facilitar su revisión visual. Las diferencias no representan necesariamente errores ni detalles inventados.

Por su parte, **SigLIP** compara las características visuales de la fotografía original y la restaurada para calcular una puntuación aproximada de similitud. Esto permite evaluar si ambas imágenes mantienen un parecido general después del proceso de restauración.

Los resultados se presentan mediante un informe sencillo que facilita su interpretación.

Es importante aclarar que **una similitud elevada no garantiza que todos los rasgos del animal se hayan conservado**. Por eso, se recomienda revisar especialmente los ojos, las orejas, el hocico y el pelaje, ya que la IA puede generar detalles que no estaban presentes en la fotografía original.

````python


# ============================================================
# PASO 3.5 - MAPA DE DIFERENCIAS Y EVALUACIÓN CON SIGLIP
# ============================================================

import cv2
import numpy as np
import torch
import torch.nn.functional as F
from PIL import Image

# ------------------------------------------------------------
# 1. GENERAR EL MAPA DE DIFERENCIAS
# ------------------------------------------------------------

def generar_mapa_diferencias(bicubica, restaurada):

    # Igualar dimensiones para comparar ambas imágenes.
    bicubica = bicubica.resize(
        restaurada.size,
        Image.Resampling.LANCZOS
    )

    gris_bicubica = cv2.cvtColor(
        np.array(bicubica.convert("RGB")),
        cv2.COLOR_RGB2GRAY
    )

    gris_restaurada = cv2.cvtColor(
        np.array(restaurada.convert("RGB")),
        cv2.COLOR_RGB2GRAY
    )

    # Detectar los detalles mediante el operador Laplaciano.
    detalle_bicubica = np.abs(
        cv2.Laplacian(gris_bicubica, cv2.CV_32F)
    )

    detalle_restaurada = np.abs(
        cv2.Laplacian(gris_restaurada, cv2.CV_32F)
    )

    # Identificar dónde aumentó el detalle.
    incremento = np.maximum(
        detalle_restaurada - detalle_bicubica,
        0
    )

    umbral = max(
        5,
        np.percentile(incremento, 95)
    )

    mascara = (
        incremento > umbral
    ).astype(np.uint8)

    mascara = cv2.dilate(
        mascara,
        np.ones((3, 3), np.uint8),
        iterations=1
    )

    # Resaltar en rojo las zonas con mayor incremento de detalle.
    resultado = np.array(
        restaurada.convert("RGB")
    ).copy()

    resultado[mascara > 0] = (
        0.5 * resultado[mascara > 0]
        + 0.5 * np.array([255, 0, 0])
    ).astype(np.uint8)

    return Image.fromarray(resultado)


# ------------------------------------------------------------
# 2. EVALUACIÓN DE SIMILITUD VISUAL MEDIANTE SIGLIP
# ------------------------------------------------------------

def evaluar_siglip(original, restaurada):

    # Preparar ambas fotografías para el modelo.
    entradas_original = procesador_visual(
        images=original.convert("RGB"),
        return_tensors="pt"
    )

    entradas_restaurada = procesador_visual(
        images=restaurada.convert("RGB"),
        return_tensors="pt"
    )

    # Obtener el dispositivo donde está cargado SigLIP.
    dispositivo = next(siglip.parameters()).device

    entradas_original = {
        clave: valor.to(dispositivo)
        for clave, valor in entradas_original.items()
        if clave == "pixel_values"
    }

    entradas_restaurada = {
        clave: valor.to(dispositivo)
        for clave, valor in entradas_restaurada.items()
        if clave == "pixel_values"
    }

    # Extraer las características visuales de ambas imágenes.
    with torch.inference_mode():

        caracteristicas_original = siglip.get_image_features(
            pixel_values=entradas_original["pixel_values"]
        )

        caracteristicas_restaurada = siglip.get_image_features(
            pixel_values=entradas_restaurada["pixel_values"]
        )

        # Algunas versiones de Transformers devuelven un objeto
        # BaseModelOutputWithPooling en lugar de un tensor.
        # Extraer el vector visual correspondiente.
        if not isinstance(caracteristicas_original, torch.Tensor):

            if hasattr(caracteristicas_original, "pooler_output"):
                caracteristicas_original = (
                    caracteristicas_original.pooler_output
                )

            elif hasattr(caracteristicas_original, "image_embeds"):
                caracteristicas_original = (
                    caracteristicas_original.image_embeds
                )

            else:
                raise ValueError(
                    "No se pudieron extraer las características "
                    "visuales de la fotografía original."
                )

        if not isinstance(caracteristicas_restaurada, torch.Tensor):

            if hasattr(caracteristicas_restaurada, "pooler_output"):
                caracteristicas_restaurada = (
                    caracteristicas_restaurada.pooler_output
                )

            elif hasattr(caracteristicas_restaurada, "image_embeds"):
                caracteristicas_restaurada = (
                    caracteristicas_restaurada.image_embeds
                )

            else:
                raise ValueError(
                    "No se pudieron extraer las características "
                    "visuales de la fotografía restaurada."
                )

        # Normalizar los vectores visuales.
        caracteristicas_original = F.normalize(
            caracteristicas_original.float(),
            dim=-1
        )

        caracteristicas_restaurada = F.normalize(
            caracteristicas_restaurada.float(),
            dim=-1
        )

        # Calcular la similitud del coseno.
        similitud = F.cosine_similarity(
            caracteristicas_original,
            caracteristicas_restaurada,
            dim=-1
        ).item()

    # Mostrar la puntuación en una escala de 0 a 100.
    # No representa el porcentaje de detalles reales.
    puntuacion = max(
        0.0,
        min(100.0, similitud * 100)
    )

    # --------------------------------------------------------
    # 3. GENERAR EL INFORME EN LENGUAJE NATURAL
    # --------------------------------------------------------

    texto = (
        "### ¿Qué tanto se parece la fotografía "
        "restaurada a la original?\n\n"
    )

    texto += (
        f"**Similitud visual aproximada: "
        f"{puntuacion:.1f} sobre 100**\n\n"
    )

    if similitud >= 0.90:

        texto += (
            "Las dos fotografías presentan un parecido "
            "visual elevado según SigLIP.\n\n"
        )

    elif similitud >= 0.75:

        texto += (
            "Las fotografías mantienen varias características "
            "visuales similares, aunque también pueden "
            "presentar diferencias.\n\n"
        )

    else:

        texto += (
            "SigLIP detectó un parecido visual menor "
            "entre ambas fotografías.\n\n"
        )

    texto += (
        "**¿Qué significa?**\n\n"
        "Esta puntuación permite comparar el parecido general "
        "entre las fotografías, pero no confirma que todos "
        "los rasgos del animal se hayan conservado.\n\n"
        "**¿Qué deberías revisar?**\n\n"
        "Compará los ojos, las orejas, el hocico "
        "y el pelaje con la fotografía original.\n\n"
        "**Importante:** La similitud es una estimación, "
        "no un porcentaje de autenticidad. "
        "La IA puede generar detalles que no estaban "
        "presentes en la fotografía original."
    )

    return texto



````

## 3.6 Integración y ejecución de la aplicación web

Los resultados se presentan mediante tres imágenes ubicadas una al lado de la otra: fotografía original, interpolación bicúbica y resultado final restaurado.

Debajo de las imágenes se muestran el mapa de diferencias y el informe de análisis visual generado mediante SigLIP, acompañados de advertencias sobre la posibilidad de que la inteligencia artificial reconstruya detalles que no coincidan con los rasgos originales del animal.

La persona usuaria puede descargar directamente la fotografía final en formato PNG, sin necesidad de seleccionar entre distintas versiones.

La interfaz está diseñada para facilitar su utilización por parte de los voluntarios del refugio, sin necesidad de ejecutar manualmente los procedimientos de procesamiento de imágenes.

````python

# ============================================================
# PASO 3.6 - INTERFAZ FINAL DE GRADIO
# ============================================================

import gradio as gr
import os
import tempfile

from PIL import Image


# ------------------------------------------------------------
# 1. VERIFICAR EL UMBRAL MÍNIMO DE RESOLUCIÓN
# ------------------------------------------------------------

UMBRAL_MINIMO = 112


def verificar_resolucion(foto):

    if foto is None:
        return (
            "Cargá una fotografía para verificar "
            "su resolución."
        )

    ancho, alto = foto.size
    lado_menor = min(ancho, alto)

    if lado_menor < UMBRAL_MINIMO:

        return (
            "### ⚠️ Advertencia: resolución inferior "
            "a la recomendada\n\n"
            f"**Resolución de la fotografía:** "
            f"{ancho} × {alto} píxeles.\n\n"
            "**Resolución mínima recomendada:** "
            "112 píxeles en el lado más corto.\n\n"
            "Las pruebas experimentales mostraron que "
            "las fotografías con resoluciones inferiores "
            "pueden presentar mayores alteraciones en "
            "los ojos, el hocico, el pelaje y otros rasgos "
            "del animal.\n\n"
            "**Podés continuar con la restauración**, "
            "pero el resultado podría incluir detalles "
            "que no estaban presentes en la fotografía original."
        )

    return (
        "### ✅ Resolución mínima recomendada cumplida\n\n"
        f"**Resolución de la fotografía:** "
        f"{ancho} × {alto} píxeles.\n\n"
        "La fotografía cumple con el umbral mínimo "
        "recomendado de **112 píxeles en el lado más corto**.\n\n"
        "Esto no garantiza que la restauración sea exacta. "
        "Stable Diffusion puede modificar algunos detalles "
        "incluso cuando se cumple el umbral."
    )



# ------------------------------------------------------------
# 2. INTEGRAR RESTAURACIÓN Y ANÁLISIS
# ------------------------------------------------------------

def procesar_fotografia(foto):

    if foto is None:
        raise gr.Error("Primero cargá una fotografía.")

    # Obtener las tres imágenes de la restauración.
    original, bicubica, sin_ruido = restaurar_fotografia(
        foto
    )

    # Generar el mapa de diferencias utilizando
    # la fotografía final procesada con OpenCV.
    mapa = generar_mapa_diferencias(
        bicubica,
        sin_ruido
    )

    # Evaluar la similitud entre la fotografía original
    # y el resultado final con Stable Diffusion + OpenCV.
    informe_siglip = evaluar_siglip(
        original,
        sin_ruido
    )

    # Devolver las cinco salidas a Gradio.
    return (
        original,
        bicubica,
        sin_ruido,
        mapa,
        informe_siglip
    )



# ------------------------------------------------------------
# 3. PREPARAR LA DESCARGA DE LA FOTOGRAFÍA FINAL
# ------------------------------------------------------------

def preparar_descarga(imagen_final):

    if imagen_final is None:
        raise gr.Error(
            "Primero restaurá una fotografía antes de descargarla."
        )

    # Crear un archivo temporal único.
    with tempfile.NamedTemporaryFile(
        suffix=".png",
        prefix="fotografia_restaurada_final_",
        delete=False
    ) as archivo:
        ruta = archivo.name

    # Guardar solamente la fotografía final.
    if isinstance(imagen_final, Image.Image):

        imagen_final.save(
            ruta,
            format="PNG"
        )

    elif isinstance(imagen_final, str):

        with Image.open(imagen_final) as imagen:
            imagen.save(
                ruta,
                format="PNG"
            )

    else:

        Image.fromarray(
            imagen_final
        ).save(
            ruta,
            format="PNG"
        )

    return ruta


# ------------------------------------------------------------
# 4. CONSTRUIR LA INTERFAZ
# ------------------------------------------------------------

with gr.Blocks(
    title="Restauración de fotografías de perros"
) as demo:


    gr.Markdown(
    "# 🐶 Restauración de fotografías de perros\n\n"
    "Ampliá fotografías pequeñas, mejorá su calidad "
    "y compará los resultados para revisar "
    "si se conservaron los rasgos del animal.\n\n"
    "**Formatos recomendados:** JPG, JPEG y PNG.\n\n"
    "**Resolución recomendada:** al menos 112 píxeles "
    "en el lado más corto de la fotografía.\n\n"
    "Las fotografías con una resolución inferior "
    "también pueden procesarse, pero podrían presentar "
    "mayores alteraciones en los detalles del animal."
    )

    entrada_foto = gr.Image(
        label="Cargar fotografía",
        type="pil",
        height=250
    )

    # --------------------------------------------------------
    # VERIFICACIÓN DE RESOLUCIÓN
    # --------------------------------------------------------

    salida_resolucion = gr.Markdown(
        value=(
            "Cargá una fotografía para verificar "
            "si cumple con la resolución mínima recomendada."
        )
    )

    boton_restaurar = gr.Button(
        "Restaurar fotografía",
        variant="primary"
    )

    # --------------------------------------------------------
    # RESULTADOS DE LA RESTAURACIÓN
    # --------------------------------------------------------

    gr.Markdown("## Resultados de la restauración")

    # Tres fotografías en una única fila.
    with gr.Row():

        salida_original = gr.Image(
            label="Fotografía original",
            type="pil",
            height=300,
            interactive=False
        )

        salida_bicubica = gr.Image(
            label="Interpolación bicúbica",
            type="pil",
            height=300,
            interactive=False
        )

        salida_final = gr.Image(
            label="Resultado final: Stable Diffusion x4 + OpenCV",
            type="pil",
            height=300,
            interactive=False
        )

    # --------------------------------------------------------
    # MAPA DE DIFERENCIAS Y ANÁLISIS VISUAL
    # --------------------------------------------------------

    gr.Markdown("## Análisis de la fotografía restaurada")

    with gr.Row():

        with gr.Column(scale=1):

            gr.Markdown("### Mapa de diferencias")

            salida_mapa = gr.Image(
                label="Zonas con mayor aumento de detalle",
                type="pil",
                height=250
            )

            gr.Markdown(
                "Las zonas rojas muestran dónde se produjeron "
                "los mayores aumentos de detalle respecto "
                "de la ampliación bicúbica de referencia.\n\n"
                "Estos cambios no necesariamente representan errores."
            )

        with gr.Column(scale=1):

            salida_siglip = gr.Markdown(
                value=(
                    "### ¿Qué tanto se parece la fotografía "
                    "restaurada a la original?\n\n"
                    "El análisis aparecerá después "
                    "de restaurar la fotografía."
                )
            )

    # --------------------------------------------------------
    # ADVERTENCIA SOBRE LA RESTAURACIÓN
    # --------------------------------------------------------

    gr.Markdown("## ¿Se conservaron los rasgos del animal?")

    gr.Markdown(
        "⚠️ **Importante sobre tu fotografía**\n\n"
        "Stable Diffusion x4 genera nuevos detalles "
        "para ampliar la imagen. Algunos pueden no coincidir "
        "exactamente con los rasgos originales del animal.\n\n"
        "Revisá especialmente los ojos, las orejas, "
        "el hocico y el pelaje.\n\n"
        "La similitud calculada por SigLIP es aproximada. "
        "No representa el porcentaje de detalles reales "
        "ni permite confirmar si la IA inventó características."
    )

    # --------------------------------------------------------
    # DESCARGA DE LA FOTOGRAFÍA FINAL
    # --------------------------------------------------------

    gr.Markdown("## Descargar fotografía restaurada")

    gr.Markdown(
        "Descargá únicamente la fotografía final "
        "procesada con Stable Diffusion x4 y OpenCV, "
        "en formato PNG y resolución completa."
    )

    boton_descargar = gr.DownloadButton(
        label="Descargar fotografía final",
        variant="primary"
    )

    # --------------------------------------------------------
    # 5. CONECTAR LAS FUNCIONES
    # --------------------------------------------------------

    # Verificar automáticamente la resolución.
    entrada_foto.change(
        fn=verificar_resolucion,
        inputs=entrada_foto,
        outputs=salida_resolucion
    )

    # Restaurar y mostrar tres imágenes, mapa y SigLIP.
    boton_restaurar.click(
        fn=procesar_fotografia,
        inputs=entrada_foto,
        outputs=[
            salida_original,
            salida_bicubica,
            salida_final,
            salida_mapa,
            salida_siglip
        ]
    )

    # Descargar únicamente el resultado final.
    boton_descargar.click(
        fn=preparar_descarga,
        inputs=[salida_final],
        outputs=boton_descargar
    )


# ------------------------------------------------------------
# 6. EJECUTAR LA APLICACIÓN
# ------------------------------------------------------------

demo.queue(default_concurrency_limit=1)
demo.launch(share=True)










````

## Conclusión Final

El desarrollo del proyecto **Restauración de Fotografías con Inteligencia Artificial** permitió integrar técnicas tradicionales de procesamiento digital de imágenes con modelos de inteligencia artificial generativa.

Se trabajó con **Stable Diffusion x4 Upscaler** para la superresolución de fotografías, utilizando la interpolación bicúbica como método de comparación. Además, se incorporaron herramientas de análisis como **SigLIP, PSNR, SSIM y mapas de diferencias**, con el objetivo de evaluar los resultados desde distintas perspectivas.

La aplicación desarrollada con **Gradio** busca facilitar el uso de estas herramientas para personas que colaboran en refugios y necesitan mejorar fotografías de perros rescatados o extraviados.

Uno de los aprendizajes más importantes fue comprender que **aumentar la nitidez de una imagen no significa necesariamente recuperar información real**. Los modelos de difusión pueden generar detalles visualmente convincentes que no estaban presentes en la fotografía original.

Por este motivo, el proyecto destaca la importancia de comparar las imágenes, informar las limitaciones del proceso y mantener la supervisión humana, especialmente cuando las características físicas del animal son importantes para su identificación.

En conclusión, la inteligencia artificial puede ser una herramienta útil para mejorar fotografías, siempre que sus resultados se interpreten críticamente y se distinga entre la información conservada y los detalles reconstruidos por el modelo.
